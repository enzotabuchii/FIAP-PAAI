```python
client = OpenAI()

combined_container = client.containers.create(name="combined_data_for_agent")

with open("contatos_pix.csv", "rb") as arquivo:
    client.containers.files.create(container_id=combined_container.id, file=arquivo)
with open("saldo.csv", "rb") as arquivo:
    client.containers.files.create(container_id=combined_container.id, file=arquivo)
```

```python
import sys

agente_contatos = Agent(
    name="Passador de informações (Contatos)",
    instructions=(
        "Use Python e pandas para analisar os dados de contatos_pix.csv e saldo.csv"
        "Responda em português brasileiro, com linguagem clara informações sobre os contatos"
        "Não se esqueça de responder a pergunta principal do usuário"
    ),
    tools=[
        CodeInterpreterTool(
            tool_config={ "type": "code_interpreter", "container": combined_container.id }
        )
    ],
    model="gpt-4o-mini",
)

query = input('Pergunte: ')

resultado = await Runner.run(
    agente_contatos,
)

print(resultado.final_output)
```
